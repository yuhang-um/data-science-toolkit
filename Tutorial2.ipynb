```python
# 1. Create a numpy array of size 2 x 2

import numpy as np

array_2x2 = np.array([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
# 2. Create a random numpy array of size 3 x 4

#The first number is the
random3x4 = np.random.rand(3,4)
print("Random numpy 3x4 array")
print(random3x4)
```

```python
# 3. Create a numpy array of ones and a numpy array of zeroes

ones_array=np.ones((3,3))
zeros_array=np.zeros((3,3))

print("Arrays of ones: \n", ones_array)
print("Arrays of zeros: \n", zeros_array)
```

```python
# 4. Create a random numpy array of size 6x3 and transpose it

random6x3=np.random.rand(6,3)
print("Original array: \n", random6x3)

transpose_array=random6x3.transpose()
print("\nTranspose array: \n", transpose_array)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot product
# !! The number of columns in the first array must equal to the number of rows in the second array

n,p,m=3,4,5

arrayA=np.random.rand(n,p)
arrayB=np.random.rand(p,m)
dot_product=np.dot(arrayA,arrayB)

print("The first array:\n", arrayA)
print("\nThe second array:\n", arrayB)
print("\nThe dot product of two random arrays:")
print(dot_product)
```
